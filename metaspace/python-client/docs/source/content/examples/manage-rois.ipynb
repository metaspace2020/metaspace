# Manage ROIs

This notebook illustrates how to export a dataset's ROIs (Regions of Interest, drawn or previously imported in the [METASPACE](https://metaspace2020.org) webapp's ion image viewer) to a GeoJSON file, validate a GeoJSON file before importing it, and import ROIs from a GeoJSON file onto a dataset.

ROI coordinates are stored in ion-image pixel space (column, row), not a geographic coordinate system. Importing a file onto a dataset whose ion-image grid doesn't match the one it was exported from -- or whose coordinates don't fit inside the target grid -- is rejected with an explanation rather than partially imported.

## Setup

```python
from metaspace import SMInstance
```

```python
sm = SMInstance()
```

```python
# This will prompt you to enter your API key if needed and it will save it to a config file.
# Note that API keys should be kept secret like passwords.
sm.save_login()
```

## Choose a dataset

```python
dsid = "2016-09-22_11h16m17s"
```

## Export ROIs to a GeoJSON file

```python
feature_collection = sm.export_rois(dsid, path="rois.geojson")
feature_collection
```

## Validate a GeoJSON file before importing it

This checks the file against the target dataset's ion-image grid without writing anything -- useful before importing a large batch, or when the file didn't come from METASPACE's own export.

```python
result = sm.validate_roi_geojson(dsid, "rois.geojson")
result
```

```python
if not result["valid"]:
    for error in result["errors"]:
        print(error["message"])
```

## Import ROIs from a GeoJSON file

Importing always appends new ROIs -- it never replaces or removes ROIs the dataset already has.

```python
imported = sm.import_rois(dsid, "rois.geojson")
imported
```
